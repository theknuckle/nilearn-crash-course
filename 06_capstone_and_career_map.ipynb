## Where nilearn shows up
| Setting | Typical use |
|---|---|
| Academic cognitive neuroscience | GLM activation maps, connectivity, decoding; figures in papers |
| Clinical research / psychiatry | connectivity-based biomarkers, group comparisons, prediction models |
| Pharma / biotech imaging | feature extraction pipelines for trials, QC and reporting |
| Neurotech / BCI | decoding pipelines (with caveats on real-time use) |
| Data science / ML | applying scikit-learn to high-dimensional imaging data |
| Open-science infrastructure | nilearn underpins reports in tools like fMRIPrep / BIDS Apps (look at `nilearn.interfaces`, `nilearn.reporting`) |

## Habits worth taking away
- Always visualise intermediate results (mask, atlas alignment, design matrix).
- Baselines + correct CV before trusting any ML number.
- Record versions (`nilearn.__version__`) and parameters; analysis choices are part of the result.
- Datasets/APIs change between nilearn versions: check docstrings (`?function`) and the changelog when a gallery example doesn't run as-is.
